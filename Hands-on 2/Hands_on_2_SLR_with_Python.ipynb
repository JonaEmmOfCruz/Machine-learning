<h1><strong>Simple Linear Regression</strong></h1>

<h3><strong>¿Qué es la regresión lineal simple?</strong></h3>

Es un modelo de regresión lineal con una sola variable explicativa. <br>
En la regresión lineal simple, predecimos puntuaciones en una variable basándonos en los resultados de otra. La variable criterio $Y$ es la variable que estamos prediciendo. La variable predictora $X$ es la variable que utilizamos para realizar nuestras predicciones. El enfoque de predicción se conoce como regresión simple porque solo hay una variable predictora.

<br>

<h3><strong>Ecuación: $y = mx + c$</strong></h3>

Esta es la ecuación de regresión lineal simple, donde $c$ es la constante y $m$ es la pendiente , y describe la relación entre $x$ (variable independiente) e $y$ (variable dependiente). <br><br>
El coeficiente puede ser positivo o negativo e indica el grado de cambio en la variable dependiente por cada unidad de cambio en la variable independiente.

> $y_{i} = \beta_{0} + \beta_{1}X_{i}$

<br><br>


<h3><strong>Implementar la regresión lineal simple en Python</strong></h3>


<h5><strong>Paso 1: Importar los paquetes de Python necesarios
</strong></h5>

```python
# Importar bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from pandas.core.common import random_state
from sklearn.linear_model import LinearRegression
from google.colab import drive
```

<h5><strong>Paso 2: Cargar el conjunto de datos
</strong></h5>

```python
# 1. Montar tu Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Definir la ruta del archivo
ruta = '/content/drive/MyDrive/Colab Notebooks/dataset/Salary_Data.csv'

# 3. Leer el archivo CSV
df = pd.read_csv(ruta)

# 4. Mostrar las primeras filas (ahora verás las 2 columnas)
df.head()
```

<h5><strong>Paso 3: Análisis de datos
</strong></h5>

```python
# Describir los datos
df.describe()
```

<h5><strong>Grafica de distribución
</strong></h5>

```python
# Distribución de datos
plt.title( 'Gráfico de distribución salarial' )
sns.distplot(df[ 'Salary' ])
plt.show()
```

<h5><strong>Verificación de la relación entre salario y experiencia.
</strong></h5>

```python
# Relación entre salario y experiencia
plt.scatter(df['YearsExperience'], df['Salary'], color='lightcoral')
plt.title('Salario vs. Experiencia')
plt.xlabel('Años de experiencia')
plt.ylabel('Salario')
plt.box(False)
plt.show()
```

<h5><strong>Paso 4: Dividir el conjunto de datos en variables dependientes e independientes.
</strong></h5>

```python
# Dividir variables
X = df.iloc[:, :1]   # Variable independiente (mantiene formato de tabla/2D)
y = df.iloc[:, 1]    # Variable dependiente (vector/1D)
```

<h5><strong>Dividir los datos en conjuntos de entrenamiento y prueba.
</strong></h5>

```python
# Dividir el conjunto de datos en prueba/entrenamiento
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2 , random_state = 0 )
```

<h5><strong>Paso 5: Entrenar el modelo de regresión
</strong></h5>

```python
# Modelo de regresión
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

<h5><strong>Paso 6: Predecir el resultado
</strong></h5>

```python
# Resultado de la predicción
y_pred_test = regressor.predict(X_test)      # Valor predicho de y_test
y_pred_train = regressor.predict(X_train)    # Valor predicho de y_train
```

<h5><strong>Paso 7: Graficar los resultados del entrenamiento y las pruebas.
</strong></h5>

```python
# Predicción en el conjunto de entrenamiento
plt.scatter(X_train, y_train, color = 'lightcoral' )
plt.plot(X_train, y_pred_train, color = 'firebrick' )
plt.title( 'Salario vs Experiencia (Conjunto de entrenamiento)' )
plt.xlabel( 'Años de experiencia' )
plt.ylabel( 'Salario' )
plt.legend([ 'X_train/Pred(y_test)' , 'X_train/y_train' ], title = 'Sal/Exp' , loc= 'best' , facecolor= 'white' )
plt.box( False )
plt.show()
```

<h5><strong>Grafica de los datos del conjunto de prueba frente a las predicciones
</strong></h5>

```python
# Predicción en el conjunto de prueba
plt.scatter(X_test, y_test, color = 'lightcoral' )
plt.plot(X_train, y_pred_train, color = 'firebrick' )
plt.title( 'Salario vs Experiencia (Conjunto de prueba)' )
plt.xlabel( 'Años de experiencia' )
plt.ylabel( 'Salario' )
plt.legend([ 'X_train/Pred(y_test)' , 'X_train/y_train' ], title = 'Sal/Exp' , loc= 'best' , facecolor= 'white' )
plt.box( False )
plt.show()
```

<h3><strong>Resultados de la regresión
</strong></h3>

```python
# Coeficientes del regresor e intercepto
print ( f'Coeficiente: {regressor.coef_} ' )
print ( f'Intercepto: {regressor.intercept_} ' )
```
